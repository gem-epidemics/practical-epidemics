## Theory

The Metropolis-Hastings (MH) algorithm is one of the most commonly used MCMC methods. It's a rejection sampling algorithm used to generate samples from distributions that are otherwise difficult to sample from directly.

Parameter estimation via the MH algorithm works by firstly drawing an initial parameter value from a prior distribution. We then propose a new parameter value by drawing from a chosen proposal distribution. We now calculate the posterior (log likelihood and log prior) for both parameter values, and calculate the accpetance probability as defined below. These values are stored as part of the MCMC chain. We now have an updated parameter value and can propose a new parameter value until we have run our MCMC for the number of samples specified.


The algorithm is as follows:

```{prf:algorithm} Metropolis-Hastings algorithm for sampling from a target distribution $\pi(\cdot)$ and proposal distribution $q(\cdot)$.

**Input:**  $q(\cdot)$ proposal distribution, $\pi(\cdot)$ a target distribution

**Initialise:** $\theta^{(0)}$,  $i=0$, $n$ number of iterations

1. **While** $i \le n - 1$  **do**

    1. Propose $\theta^* \sim q(\cdot | \theta^{(i)})$
    2. Compute acceptance probability $\alpha(\theta^{(i)}, \theta^*) = 1  \wedge \frac{\pi(\theta^*)q(\theta^{(i)} | \theta^*)}{\pi(\theta^{(i)})q(\theta^{*} | \theta^{(i)})}$
    3. **If** $\mbox{Uniform}(0,1) < \alpha(\theta^{(i)}, \theta^*)$ **then**
    $\theta^{(i+1)} = \theta^*$
    **else** $\theta^{(i+1)} = \theta^{(i)}$  
    4. $i ← i + 1$
```

**Symmetric proposals**

If the specified proposal distribution is symmetric the acceptance ratio simplifies.

Symmetric proposal distributions are those which satisfy $q(\theta^{(i)} | \theta^*) = q(\theta^{*} | \theta^{(i)})$. Gaussian and Cauchy distributions centered on the current state are examples of symmetric proposals.

In the symmetric case, the 'q ratio' cancels, as $\frac{q(\theta^{(i)} | \theta^*)}{q(\theta^{*} | \theta^{(i)})} = 1.$

The acceptance probability is therefore simplified to $\alpha(\theta^{(i)}, \theta^*) = 1 \wedge \frac{\pi(\theta^*)}{\pi(\theta^(i)}$.

### Specify Bayesian model

Specify a simple homoegeneous mixing SIR epidemic model using `JointDistributionCoroutineAutoBatched`. We are interested in estimating the infectious rate $\beta$, and therefore need to specify a prior distribution.

`````python
@tfd.JointDistributionCoroutineAutoBatched
def model():

  # Prior
  beta = yield tfd.Gamma(
      concentration=DTYPE(1.0), rate=DTYPE(100.0), name="beta",
  )

  # Epidemic model
  incidence_matrix = np.array(
    [ #  SI  IR
        [-1,  0],  # S
        [ 1, -1],  # I
        [ 0,  1], # R
    ],
    dtype=DTYPE
  )

  initial_state = np.array([[999, 1, 0]]).astype(DTYPE)

  def si_rate(t, state):
    return beta * state[:, 1]
    
  def ir_rate(t, state):
    return 0.1

  sir =  yield DiscreteTimeStateTransitionModel(transition_rate_fn=[si_rate, ir_rate],
                                      incidence_matrix=incidence_matrix,
                                      initial_state=initial_state,
                                      initial_step=0,
                                      time_delta=1.0,
                                      num_steps=100,
                                      name ="sir")
`````

Run a simulation from the prior distribution.

`````python
seed = jax.random.key(0)
beta_val, sim_vals = model.sample(seed=seed)
`````

Condition the model on having some epidemic output, in this case simulation data (sim_vals).

`````python
conditioned_model = model.experimental_pin(sir=sim_vals)
`````

`````python
def mh_accept(proposed_log_prob, current_log_prob, log_q_ratio, seed):
  """Decides whether to accept or reject a MH move"""
  log_acceptance_prob = proposed_log_prob - current_log_prob + log_q_ratio
  u = tfd.Bernoulli(probs=jnp.exp(log_acceptance_prob), dtype=bool).sample(seed=seed)

  return u


def target_log_prob_fn(beta_val):
    return conditioned_model.log_prob(beta=beta_val)


def mh_algo(
    target_log_prob_fn,
    proposal_distribution_scale,
    init_state,
    num_samples,
    seed=0,
):
    seeds = jax.random.split(jax.random.key(seed), num=num_samples)
    
    samples = []
    accept = []

    #set current state and compute log prob
    current_state = init_state
    current_log_prob = target_log_prob_fn(current_state)

    for i in range(num_samples):

      proposal_seed, accept_seed = jax.random.split(seeds[i])
    
      # Compute current likelihood
      current_log_prob = target_log_prob_fn(current_state)

      # (Forward) proposal distribution
      proposal_distn = tfp.distributions.LogNormal(
          loc=jnp.log(current_state), scale = proposal_distribution_scale
      )
    
      # Draw proposal
      proposed_state = proposal_distn.sample(seed=proposal_seed)
    
      # Reverse proposal distribution
      rev_proposal_distn = tfp.distributions.LogNormal(
          loc=jnp.log(proposed_state), scale = proposal_distribution_scale
      )

      # Compute proposed log likelihood
      proposed_log_prob = target_log_prob_fn(proposed_state)

      # Calculate log proposal density ratio
      log_q_fwd = proposal_distn.log_prob(proposed_state)
      log_q_rev = rev_proposal_distn.log_prob(current_state)
      log_q_ratio = log_q_rev - log_q_fwd

      # Accept or reject?
      is_accepted = mh_accept(
          proposed_log_prob,
          current_log_prob,
          log_q_ratio,
          accept_seed,
      )
      if is_accepted:
        current_state = proposed_state

      samples.append(current_state)
      accept.append(is_accepted)

    return np.stack(samples, axis=0), np.stack(accept, axis=0)


num_samples = 1000

samples, accept = mh_algo(target_log_prob_fn=target_log_prob_fn,
    proposal_distribution_scale=0.1,
    init_state=0.0025,
    num_samples=num_samples)
`````

Using `gemlib` and `jax.jit` we can run a Metropolis-Hastings algorithm for many more samples much more quickly.

`````python
from gemlib.mcmc import mcmc, rwmh

kernel = rwmh(scale=0.0007)

# Wrap a sample function for JIT compilation
@jax.jit
def run_mcmc(initial_position, seed):
    return mcmc(
        num_samples=10000,
        sampling_algorithm=kernel,
        target_density_fn=conditioned_model.log_prob,
        initial_position=initial_position,
        seed=seed,
)

# sample chain and store results
samples_rw, info = run_mcmc(
    initial_position=DTYPE(0.0025),
    seed=jax.random.key(1)
)

plt.plot(samples_rw[1000:]) # burn in
plt.hlines(xmin=0, xmax=samples_rw[1000:].shape[0], y=beta_val, color="red")
plt.title(f'Trace plot of Beta, (acceptance {np.mean(info.is_accepted)*100}%)')
`````
