### Basics 

#### Arithmetic operators
These are used for basic arithmetic. 
| Operator | Name | Description | Example |

| `**` | Exponentiation | Raises a number to a power | `5 ** 2` → `25` |

#### Comparison operators
These are used for performing boolean (true/false) evaluations. 
| Operator | Name | Description | Example |
|----------|------|-------------|---------|

#### `datetime`
Epidemiological data is almost always timestamped, so being able to do arithmetic on dates is essential (e.g. incubation periods, time-to-recovery). The `datetime` module lets us parse a date string into a `datetime` object using `datetime.strptime`, and once we have `datetime` objects, subtracting one from another gives us a `timedelta` representing the elapsed time between them. These are special types that provide convenience. 

`````python
from datetime import datetime

infected = datetime.strptime(epi_case["infected_date"], "%Y-%m-%d")
recovered = datetime.strptime(epi_case["recovery_date"], "%Y-%m-%d")

recovery_time = recovered - infected
print(type(recovery_time))
print(f"{epi_case['name']} took {recovery_time.days} days to recover")
`````

We can also add a `timedelta` back onto a `datetime` to project forward in time, e.g. scheduling a follow-up appointment 14 days after recovery:

`````python
from datetime import timedelta

follow_up = recovered + timedelta(days=14)
print(f"Follow-up appointment: {follow_up.strftime('%Y-%m-%d')}")
`````

#### `float32` vs `float64`
`numpy` gives us numeric types with a fixed number of bits, which controls both the *precision* of a value (how many significant digits it can represent) and its memory footprint. `float64` (double precision) uses 64 bits and is numpy's default; `float32` (single precision) uses half the memory, but stores fewer significant digits, which can introduce rounding error into a calculation.

`````python
# same calculation, different precision
a32 = np.float32(1) / np.float32(3)
a64 = np.float64(1) / np.float64(3)

print(f"float32: {a32:.20f}")
print(f"float64: {a64:.20f}")
`````

That's the precision side of the trade-off. We can measure the speed side using the `timeit` module, which repeatedly runs a piece of code and reports how long it took.

`````python
import timeit

n = 10_000_000
x32 = np.ones(n, dtype=np.float32)
x64 = np.ones(n, dtype=np.float64)

time32 = timeit.timeit(lambda: x32 * x32, number=50)
time64 = timeit.timeit(lambda: x64 * x64, number=50)

print(f"float32: {time32:.4f}s")
print(f"float64: {time64:.4f}s")
`````

Because `float64` carries twice as many bits through every operation, it is typically slower and uses more memory than `float32` for the same computation — that's the "cost" of its extra precision. As a rule of thumb: default to `float64` for numerical stability, and only drop to `float32` when memory or speed genuinely matter and you're confident the precision loss won't affect your results.

## Arrays 
Numpy is a popular Python package that is focused on numerical computation. It is used for creating numerical arrays and provides performant implementations of many mathematical operations (e.g. mathematical functions, random number generation, linear algebra, etc.). 

By convention, the `numpy` package is imported and aliased with `np` (see above). Arrays carry less auxiliary information than data frames (more of these shortly) and so, they are faster for numerical computation. 


Accessing values, broadcasting, dimensions/rank

`````python
# We can define an array using a list:
infection_times = np.array([1.5, 3.2, 9.0, 4.0, 4.3])

# We can then access an element of the array
# NB: remember in Python structures are zero indexed
print(infection_times[0])
`````

`````python
# Defining a 2-D array
connectivity_matrix = np.array([[0, 1.5, 0.25], [1.5, 0, 0.5], [0.25, 0.5, 0]])

# We can access elements and rows/columns in the same way
print(connectivity_matrix[0, 2])
`````

Arrays can be multi-dimensional and mathematical operations can be broadcasted across the axes. The way the arrays are broadcasted comes down to the shapes. In the example below, we take a connectivity matrix that changes over time along with an array of vectors of the population at the respective time stamps. 

`````python
connectivity_matrix_over_time = np.array(
    [
        [[0, 1.5, 0.25], [1.5, 0, 0.5], [0.25, 0.5, 0]],  # time 1
        [[0, 1.7, 0.5], [1.7, 0, 0.5], [0.5, 0.5, 0]],  # time 2
        [[0, 2.0, 0.1], [2.0, 0, 0.25], [0.1, 0.25, 0]],  # time 3
        [[0, 3.0, 0.01], [3.0, 0, 0.5], [0.01, 0.5, 0]],  # time 4
    ]
)

population_count_v1 = np.array(
    [
        [100, 50, 25],  # time 1
        [125, 75, 5],  # time 2
        [250, 50, 10],  # time 3
        [300, 100, 1],  # time 4
    ],
)

# notice the difference in number of squre brackets
population_count_v2 = np.array(
    [
        [[100], [50], [25]],  # time 1
        [[125], [75], [5]],  # time 2
        [[250], [50], [10]],  # time 3
        [[300], [100], [1]],  # time 4
    ]
)
`````

`````python
population_count_v1.transpose()
`````

Let's look at the shapes of the structures we just created and then do some matrix-vector multiplication to see the different behaviours. 

`````python
print(f"Connectivity matrix shape: {connectivity_matrix.shape}")
print(f"Connectivity matricies over time shape: {connectivity_matrix_over_time.shape}")

print(f"Population vectors V1 shape: {population_count_v1.shape}")
print(f"Population vectors V2 shape: {population_count_v2.shape}")
`````

`````python
# connectivity matrix x population counts over time
# right matrix gets broadcast across time dimension - the 4
np.matmul(population_count_v1, connectivity_matrix)
`````

`````python
# connectivity matrix x population counts over time
np.matmul(connectivity_matrix, population_count_v2)
`````

`````python
# connectivity matrix x population counts over time
np.matmul(population_count_v1, connectivity_matrix_over_time)
`````

`````python
# connectivity matrix x population counts over time - notice this now produces 3 vectors each of dimension 3
np.matmul(connectivity_matrix_over_time, population_count_v2)
`````

The four `matmul` calls above all illustrate the same point: broadcasting is driven by the *trailing* (last two) axes of the arrays. `np.matmul` treats an array's final two dimensions as the matrix to multiply and treats any leading dimensions as a "batch" that gets broadcast over. `connectivity_matrix` has shape `(3, 3)` — a single matrix — while `connectivity_matrix_over_time` has shape `(3, 3, 3)`, i.e. a batch of three `(3, 3)` matrices (one per time step). Multiplying the single matrix against `population_count_v1` (shape `(3, 3)`) or `population_count_v2` (shape `(3, 3, 1)`) just performs one matrix-matrix or matrix-vector product, and the trailing dimensions (`3, 3`) x (`3, 3`) or (`3, 3`) x (`3, 1`) must be compatible for the multiplication to be defined. Multiplying the *time-varying* connectivity matrices against `population_count_v1` or `population_count_v2` broadcasts the single population array across the leading (time) axis of `connectivity_matrix_over_time`, applying the appropriate connectivity matrix to the population at each time step and returning a stack of three results. In every case, it is the shape of the trailing two axes — not the overall number of dimensions — that dictates whether the operation is valid and what shape the result takes.

## Importing and viewing data

`pandas` is the Python package that is used for creating and working with DataFrames. In this section, we look at importing data from different locations and from different types. After that, we convert them to pandas DataFrames and examine them.

### What is a DataFrame?

A **DataFrame** is `pandas`' core data structure. It is a 2-dimensional, labelled table that looks a lot like a spreadsheet. Unlike a `numpy` array, where the elements must all have the same dtype, each *column* of a DataFrame can hold a different type (e.g. names as `str`, ages as `float`, dates as `datetime`). Each value within a single column shares the same type. Each column has a name and each row has an index which is helpful for accessing values.

### What is an xarray?

An **xarray** is `xarray`' core data structure that generalized 2D `pandas` DataFrames into multi-dimensional labelled arrays. At its core, it behaves like `numpy` but adds extra functionality by being able to access (and operate) via the labels. For now, we will only show how to import a dataset with `xarray` but further examples can be found in the online [tutorials](https://docs.xarray.dev/en/stable/getting-started-guide/tutorials-and-videos.html). 

### Bringing in data from different sources

Both pakcages can read data from a **local paths** or from a **URL**. The reader function you call (`pd.read_csv`, `xr.open_dataset`, etc.) doesn't care which kind of source you give it, as long as the string you pass resolves to a valid file. Below, we load the same Hagelloch measles outbreak dataset twice: once from a file on disc, and once from a link.

`````python
# import data from disc
HAGELLOCH_DATA = xr.open_dataset("data/hagelloch.nc")

HAGELLOCH_DATA
`````

`````python
# import data from web
HAGELLOCH_DATA = (
    "https://drive.google.com/uc?export=download&id=1T-O_51EsNy2FmqnHkR_7pHZnDGC28IeC"
)

data = pd.read_csv(HAGELLOCH_DATA)

data
`````

````{tip} Does the file format matter?

`.csv` files are plain text which makes them a great default for flat, rectangular data like a line list. `.nc` (NetCDF) files are binary and self-describing and tend to read/write faster for large datasets (better management of multiple dimensions/labels).

In practice: if your data is a single flat table, `.csv` is usually the right (and most portable) choice. Consider a format like `.nc` only once your data has real multi-dimensional structure (e.g. a matrix that varies over time), or you're working at a scale where file size and read speed becomes a bottleneck.
````

### Exploring a DataFrame with pandas

Once loaded, `data` gives us the Hagelloch measles line list as a table of 188 cases (rows) and 26 recorded fields (columns). Typing the name (or letting it be the last line of a cell) displays the whole table, while DataFrame methods like `.shape` and `.columns` tells us how many rows and columns we have, and  the column names respectively:

`````python
print(data.shape)
print(data.columns)
`````

`.dtypes` shows the data type pandas inferred for each column:

`````python
data.dtypes
`````

`.head()` and `.tail()` preview just the first or last few rows — much more manageable than printing the whole table:

`````python
data.head()
`````

`.info()` gives a compact summary — column names, non-null counts, and dtypes all at once — useful for spotting missing data:

`````python
data.info()
`````

`.describe()` computes summary statistics (count, mean, std, min/max, quartiles) for every numeric column:

`````python
data.describe()
`````

Select a single column with square-bracket indexing — this returns a pandas `Series`:

`````python
data["AGE"]
`````

Select rows and columns together with `.loc` (label-based) or `.iloc` (position-based). For example, the age and sex of the first 5 cases:

`````python
# label-based accessing
data.loc[0:4, ["AGE", "SEX"]]
`````

`````python
# position-based indexing
data.iloc[0:4, [4, 5]]
`````

Finally, `.value_counts()` tallies how many times each value appears in a column — handy for a quick breakdown of a categorical field, like school class:

`````python
data["CL"].value_counts()
`````
